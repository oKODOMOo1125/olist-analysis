```python
order_path = '../data/raw/olist/olist_orders_dataset.csv'

o_guess = pd.read_csv(order_path)
print(o_guess.dtypes)
o_guess['order_purchase_timestamp'].head(3).tolist()
```

```python
date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

orders = pd.read_csv(order_path, parse_dates=date_cols)
print(orders.dtypes)
```

```python
gap = orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']

print(gap.head(3))
print(gap.dt.days.head(3).tolist())
print(gap.dt.days.median())
```

```python
print(gap.isna().sum())
print(orders['order_status'].value_counts())
```

```python
has_date = orders['order_delivered_customer_date'].notna()
pd.crosstab(orders['order_status'], has_date)
```
